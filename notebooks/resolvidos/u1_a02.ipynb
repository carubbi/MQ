Uma pergunta estatística antecipa que as respostas podem variar entre as unidades observadas. Antes de calcular qualquer resultado, é necessário identificar **quem ou o que foi observado**, **qual conjunto se pretende conhecer** e **quais informações foram registradas**.

### 3.1 O que é Estatística e como se investiga

A **Estatística** é a área do conhecimento que reúne métodos para obter, organizar, analisar e interpretar dados, considerando a variabilidade para compreender fenômenos e apoiar decisões. Ela não se limita aos cálculos: relaciona a pergunta, a obtenção dos dados, a análise e a conclusão (BARBETTA; BORNIA; REIS, seções 1.1–1.2; ARAÚJO; SILVA, seções 1.1–1.3).

Uma investigação estatística articula etapas dependentes entre si:

1. delimitar o problema;
2. formular uma pergunta compatível com os dados;
3. definir a população de interesse, a unidade de análise e a unidade observacional;
4. examinar ou planejar a obtenção dos dados;
5. analisar a variabilidade e os padrões;
6. comunicar a conclusão e explicitar seu alcance.

Uma falha nas primeiras etapas não é corrigida automaticamente por cálculos corretos nas etapas finais (ARAÚJO; SILVA, seções 1.1–1.3; BARBETTA; BORNIA; REIS, seções 1.2 e 2.1).

### 3.2 Contexto dos dados

A aplicação utiliza 344 registros do conjunto Palmer Penguins. Cada linha corresponde a um pinguim observado no contexto dos estudos documentados; as colunas registram características como espécie, ilha e massa corporal.

Para esta aula, distinguiremos:

- **população finita de referência:** as 344 linhas disponíveis no arquivo;

- **unidade observacional:** o pinguim no qual as características foram observadas e registradas;

Nesta aplicação, a unidade de análise e a unidade observacional coincidem. Em outras investigações, a informação pode ser coletada em uma unidade e a conclusão formulada sobre outra. Essa convenção permite estudar seleção e variabilidade dentro do arquivo, mas não transforma os registros disponíveis em uma representação automática de todos os pinguins das três espécies (MORDEN, 2024, seção 3.7; HORST; HILL; GORMAN, 2020).

### 4.2 Variabilidade entre amostras

A **variabilidade amostral** aparece quando amostras diferentes, obtidas pelo mesmo procedimento, produzem composições ou estatísticas diferentes. Assim, duas amostras aleatórias de dez registros podem apresentar proporções distintas de espécies e médias distintas de massa corporal.

Essa diferença é esperada em amostras pequenas e não comprova, isoladamente, a existência de viés. O notebook verificará essa ideia comparando duas seleções reproduzíveis da mesma população finita de referência.

## 5. População, amostra, representatividade e vieses

- **População de interesse:** conjunto de unidades sobre o qual se pretende responder.
- **População finita de referência:** conjunto operacional adotado para a comparação interna desta aula.
- **Quadro amostral:** representação das unidades que poderiam ser selecionadas pelo procedimento.
- **Amostra:** subconjunto efetivamente selecionado e analisado.
- **Amostragem:** procedimento usado para selecionar as unidades.
- **Amostragem aleatória simples:** procedimento no qual cada amostra possível de mesmo tamanho tem a mesma probabilidade de ser selecionada.

A população não é uma propriedade fixa do arquivo: ela depende da pergunta. Se o objetivo é descrever somente os registros Gentoo, o subconjunto Gentoo pode ser a população de interesse. Se o objetivo é comparar as três espécies, o mesmo filtro exclui grupos necessários à investigação. A seleção aleatória simples só oferece cobertura adequada quando o quadro amostral inclui as unidades elegíveis; neste notebook, o sorteio é aleatório apenas em relação às linhas disponíveis no arquivo (BARBETTA; BORNIA; REIS, seções 1.6 e 2.2.1; NAVIDI, seção 1.1; PINHEIRO et al., seção 1.2).

**Figura 2 - Relação entre população, amostra, amostragem e inferência.** A inferência parte da amostra para responder sobre uma população definida, mas sua validade depende do procedimento de seleção e das condições assumidas. Fonte: Barbetta, Bornia e Reis (2010, figura 2.2).

### 5.1 Representatividade não é sinônimo de tamanho

Uma amostra é útil quando o procedimento de seleção fornece cobertura adequada das unidades relevantes para a pergunta. Aumentar o número de observações pode reduzir a variabilidade amostral, mas não corrige automaticamente a exclusão sistemática de grupos.

Além disso, um sorteio aleatório entre as linhas disponíveis distribui a chance **dentro do arquivo**; ele não corrige limitações da coleta que produziu o próprio arquivo. Portanto, `random_state` torna o sorteio reproduzível, mas não torna a amostra representativa de uma população biológica mais ampla (BARBETTA; BORNIA; REIS, 2010, seção 2.2.1; NAVIDI, 2024, seção 1.1).

### 5.2 Variabilidade e viés não são a mesma coisa

| Fenômeno | Característica | Exemplo nesta aula |
| --- | --- | --- |
| Variabilidade amostral | Resultados mudam entre amostras obtidas pelo mesmo procedimento. | Duas amostras aleatórias de dez registros apresentam composições diferentes. |
| Viés de seleção | O procedimento favorece ou exclui sistematicamente unidades relevantes. | Filtrar somente Gentoo quando o objetivo é comparar as três espécies. |
| Viés de medição | O registro se afasta sistematicamente da característica que se pretende medir. | Deve ser considerado conceitualmente, mas não pode ser diagnosticado apenas com as saídas usadas nesta aula. |

A aleatoriedade aplicada depois de um filtro inadequado não recupera os grupos já excluídos. Por outro lado, diferenças ocasionais entre duas amostras aleatórias não bastam para concluir que houve viés.

### 5.3 Comparar amostras aleatórias reproduzíveis

Duas amostras de mesmo tamanho serão obtidas pelo mesmo procedimento, com sementes diferentes. A comparação permitirá separar diferenças esperadas entre sorteios de um mecanismo sistemático de exclusão.

```python
# Sortear 10 observações sem reposição.
amostra_a = populacao_referencia.sample(
    n=10,
    random_state=42,
)

amostra_a
```

**Tabela 5 - Amostra aleatória A.** Foram sorteadas 10 linhas da população finita de referência. O valor de `random_state` permite reproduzir o mesmo resultado (PANDAS DEVELOPMENT TEAM, 2026). Fonte dos dados: Horst, Hill e Gorman (2020).

```python
# Repetir o sorteio com outra semente.
amostra_b = populacao_referencia.sample(
    n=10,
    random_state=7,
)

amostra_b
```

**Tabela 6 - Amostra aleatória B.** A mudança de `random_state` produz outra seleção reproduzível de 10 linhas. Fonte: elaboração própria com dados de Horst, Hill e Gorman (2020).

```python
# Contar as espécies presentes nas duas amostras.
composicao_amostra_a = amostra_a["Species"].value_counts()
composicao_amostra_b = amostra_b["Species"].value_counts()

# Reunir as contagens para facilitar a comparação.
comparacao_composicoes = pd.DataFrame(
    {
        "População de referência": composicao_referencia,
        "Amostra A": composicao_amostra_a,
        "Amostra B": composicao_amostra_b,
    }
).fillna(0).astype(int)

comparacao_composicoes
```

**Tabela 7 - Contagens por espécie na população e nas amostras.** As duas amostras foram obtidas pelo mesmo procedimento, mas apresentam contagens diferentes. Como a população possui 344 observações e cada amostra possui 10, a comparação proporcional será feita na próxima tabela. Fonte: elaboração própria com dados de Horst, Hill e Gorman (2020).


```python
# Calcular as proporções de cada espécie, em porcentagem.
proporcoes_referencia = composicao_referencia / len(populacao_referencia) * 100
proporcoes_amostra_a = composicao_amostra_a / len(amostra_a) * 100
proporcoes_amostra_b = composicao_amostra_b / len(amostra_b) * 100

# Reunir e arredondar as porcentagens.
comparacao_proporcoes = pd.DataFrame(
    {
        "População de referência (%)": proporcoes_referencia,
        "Amostra A (%)": proporcoes_amostra_a,
        "Amostra B (%)": proporcoes_amostra_b,
    }
).fillna(0).round(1)

comparacao_proporcoes
```

**Tabela 8 - Proporções das espécies na população e nas amostras.** Adelie corresponde a 44,2% da população de referência, mas representa 10,0% da amostra A e 50,0% da amostra B. Gentoo passa de 36,0% na população para 60,0% na amostra A e 20,0% na amostra B. Essas diferenças ilustram a variabilidade da composição entre amostras aleatórias pequenas. Fonte: elaboração própria com dados de Horst, Hill e Gorman (2020).

**Registro do estudante:** a seleção aleatória produziu uma cópia exata da composição da população? Essas diferenças, isoladamente, comprovam viés?

**Resposta:** não. Nenhuma das amostras reproduziu exatamente a composição da população de referência. As diferenças observadas entre sorteios de dez registros são compatíveis com variabilidade amostral e, isoladamente, não comprovam viés. Para caracterizar viés, seria necessário identificar um mecanismo de seleção que favoreça ou exclua sistematicamente grupos relevantes ao objetivo.


```python
# Calcular a média da massa corporal em cada conjunto.
media_massa_referencia = populacao_referencia["Body Mass (g)"].mean()
media_massa_amostra_a = amostra_a["Body Mass (g)"].mean()
media_massa_amostra_b = amostra_b["Body Mass (g)"].mean()

# Reunir e arredondar as médias.
comparacao_medias = pd.Series(
    {
        "População de referência": media_massa_referencia,
        "Amostra A": media_massa_amostra_a,
        "Amostra B": media_massa_amostra_b,
    },
    name="Média da massa corporal (g)",
).round(1)

comparacao_medias
```

**Tabela 9 - Média da massa corporal na população e nas amostras.** A média é 4.201,8 g na população de referência, 4.262,5 g na amostra A e 3.872,5 g na amostra B. A mesma estatística assume valores diferentes porque foi calculada em seleções diferentes. Isso evidencia variabilidade amostral e não demonstra, por si só, a existência de viés. Fonte: elaboração própria com dados de Horst, Hill e Gorman (2020).

### 6.1 Parâmetro, estimador e estimativa

Um **parâmetro** $\theta$ descreve uma característica da população e, em geral, é desconhecido. Uma **estatística amostral** é uma função dos valores da amostra. Quando essa estatística é usada para estimar o parâmetro, a regra definida antes da observação dos dados é o **estimador**:

$$
\widehat{\theta}=T(X_1,X_2,\ldots,X_n).
$$

Depois que os valores são observados, obtém-se a **estimativa**, isto é, o resultado numérico do estimador:

$$
\hat{\theta}=T(x_1,x_2,\ldots,x_n).
$$

Aqui, $X_i$ representa a variável aleatória associada à unidade $i$, $x_i$ representa o valor observado e $n$ é o tamanho da amostra.

A correção do cálculo não garante a validade da generalização. Também são necessários uma população bem definida, um procedimento de seleção justificável e hipóteses compatíveis com o método inferencial (ARAÚJO; SILVA, seção 1.1; PINHEIRO et al., seção 1.1).

### 6.2 Como declarar o alcance

Uma conclusão responsável informa:

- quais observações foram analisadas;
- qual população foi definida;
- como as unidades foram selecionadas;
- quais grupos ou períodos ficaram fora da cobertura;
- se o resultado é descritivo, inferencial ou causal;
- quais limitações impedem uma generalização mais ampla.

No Palmer Penguins, uma formulação apropriada começa por “nos registros disponíveis” ou “na amostra selecionada”. A expressão “em todos os pinguins dessas espécies” exigiria justificativas que o arquivo, isoladamente, não fornece.

```python
# Calcular as contagens e as proporções observadas de cada espécie.
contagens_especies = populacao_referencia["Species"].value_counts()
proporcoes_especies = (contagens_especies / len(populacao_referencia) * 100).round(1)

# Calcular a quantidade válida, o menor e o maior valor de massa corporal.
massa_corporal = populacao_referencia["Body Mass (g)"]
amplitude_massa = pd.Series(
    {
        "Quantidade válida": massa_corporal.count(),
        "Mínimo": massa_corporal.min(),
        "Máximo": massa_corporal.max(),
    }
)

# Reunir as descrições em uma única tabela.
descricao_arquivo = pd.concat(
    [contagens_especies, proporcoes_especies, amplitude_massa],
    keys=["Contagem por espécie", "Proporção por espécie (%)", "Massa corporal (g)"],
    names=["Medida", "Categoria"],
).to_frame(name="Valor")

descricao_arquivo
```

**Tabela 13 - Descrição do arquivo por espécie e massa corporal.** As contagens, proporções e amplitudes pertencem às observações disponíveis. Fonte: elaboração própria com dados de Horst, Hill e Gorman (2020).

**Registro do estudante:** escreva uma conclusão descritiva sustentada pela saída e uma generalização que a seleção disponível não permite sustentar.

**Resposta:** no arquivo, Adelie possui 152 registros (44,2%), Gentoo 124 (36,0%) e Chinstrap 68 (19,8%). Entre os 342 valores válidos de massa corporal, os valores variam de 2.700 g a 6.300 g. Esses resultados não permitem afirmar que as mesmas proporções ou a mesma amplitude caracterizam todos os pinguins dessas espécies, pois o arquivo não é uma amostra probabilística dessa população biológica.

## 7. Síntese

- A pergunta estatística orienta a definição da população e a escolha dos dados.
- Variação entre unidades e variação entre amostras são esperadas.
- Seleção aleatória dentro de um arquivo não elimina limitações da coleta original.
- Representatividade depende do procedimento e da cobertura, não apenas do tamanho.
- Variabilidade amostral não é sinônimo de viés.
- Um filtro deve ser avaliado em relação ao objetivo da investigação.
- Descrever uma base não equivale a generalizar para uma população mais ampla.
- Afirmações causais exigem desenho e evidências compatíveis.
