## 3. Fundamentos de Estatística


### 3.1 Estatística e variabilidade

A **Estatística** reúne técnicas para coletar, organizar, descrever, analisar e interpretar dados provenientes de observações ou experimentos, contribuindo para resolver problemas e produzir conhecimento (BARBETTA; BORNIA; REIS, 2010, seção 1.1).

A **variabilidade** corresponde às diferenças encontradas entre pessoas, animais, objetos ou situações estudadas. Pinguins, por exemplo, podem apresentar massas corporais diferentes. A Estatística considera essas diferenças para compreender fenômenos e apoiar decisões (BARBETTA; BORNIA; REIS, seções 1.1–1.2).

**Figura 1 - Processo iterativo das pesquisas empíricas.** Os resultados de uma pesquisa podem originar novas perguntas. Fonte: reprodução de Barbetta, Bornia e Reis (2010, figura 1.1, p. 12; página 13 do PDF).

### 3.2 Unidades, variáveis e dados

Para descrever um conjunto de dados, distinguiremos:

- **Unidade de análise:** entidade sobre a qual se deseja formular uma descrição ou conclusão, como uma pessoa, um animal ou uma escola.
- **Unidade observacional:** entidade na qual as características são observadas ou medidas.
- **Variável:** característica investigada, como espécie ou massa corporal.
- **Dado estatístico:** valor registrado de uma variável, como uma massa de 4.200 g.
- **Observação:** conjunto do que foi observado em um caso; em uma matriz de dados, corresponde a uma linha, enquanto cada coluna corresponde a uma variável.

Essa organização permite reunir vários dados de um mesmo caso, como espécie, ilha e massa corporal (BARBETTA; BORNIA; REIS, 2010, seção 3.1).

A **informação** é extraída dos dados por meio de sua análise e interpretação em relação ao problema investigado. Organizar e examinar os dados ajuda a compreender o fenômeno e a identificar aspectos relevantes para a pergunta (PINHEIRO et al., 2009, seção 1.1).

Unidade de análise e unidade observacional podem coincidir ou diferir. Se medimos pinguins e concluímos sobre pinguins, coincidem. Se coletamos respostas de alunos para concluir sobre escolas, os alunos são as unidades observacionais e as escolas são as unidades de análise (MORDEN, 2024, seção 3.7).

### 3.3 População e amostra

A **população de interesse** é o conjunto completo de elementos cujas características se deseja estudar. Uma **amostra** é um subconjunto desses elementos, selecionado segundo regras definidas (PINHEIRO et al., 2009, seção 1.2).

Se a pergunta trata dos 344 registros disponíveis do Palmer Penguins, esse conjunto é a população de interesse; dez registros selecionados formam uma amostra. Se a pergunta muda para todos os pinguins dessas espécies, a população de interesse passa a ser mais ampla que o conjunto disponível.

Por isso, é necessário definir a população antes de selecionar a amostra.

### 3.4 Estatística Descritiva e Inferência

A **Estatística Descritiva** organiza, resume e apresenta os dados observados, por meio de tabelas, gráficos e resumos numéricos (BARBETTA; BORNIA; REIS, 2010, abertura do capítulo 3).

A **Inferência Estatística** utiliza resultados de uma amostra para formular conclusões sobre a população da qual ela foi extraída (BARBETTA; BORNIA; REIS, 2010, seção 1.1). Essa passagem envolve incerteza: outras amostras podem produzir resultados diferentes. Sua validade depende de como os dados foram obtidos e das condições exigidas pelo método utilizado.

**Exemplo:** contar os pinguins de cada espécie entre dez registros selecionados descreve essa amostra. Usar esses resultados para responder sobre os 344 registros envolve inferência. O cálculo feito na amostra, por si só, não justifica a extensão da conclusão à população.

### 4.1 Etapas da investigação

Uma investigação estatística articula etapas dependentes entre si:

1. delimitar o problema;
2. formular uma pergunta compatível com os dados;
3. definir a população de interesse, a unidade de análise e a unidade observacional;
4. examinar ou planejar a obtenção dos dados;
5. analisar a variabilidade e os padrões;
6. comunicar a conclusão e explicitar seus limites.

Uma falha nas primeiras etapas não é corrigida automaticamente por cálculos corretos nas etapas finais (BARBETTA; BORNIA; REIS, 2010, seção 2.1).

### 5.2 Variabilidade entre amostras

A **variabilidade amostral** ocorre quando amostras diferentes, obtidas pelo mesmo procedimento, produzem resultados diferentes. Duas amostras de dez registros, por exemplo, podem conter quantidades distintas de cada espécie e apresentar massas corporais médias diferentes.

Na próxima seção, definiremos o procedimento de seleção e compararemos duas amostras da população finita de referência. A comparação permitirá examinar diferenças entre seleções feitas pelo mesmo procedimento.

## 6. População, amostra, representatividade e vieses

A população de interesse depende da pergunta. Para descrever os registros das três espécies, consideramos o conjunto completo; para descrever somente Gentoo, consideramos esse subconjunto. Em cada caso, precisamos decidir quais unidades observar e como selecioná-las.

- **Censo:** levantamento de todos os elementos da população (PINHEIRO et al., 2009, seção 1.2).
- **Amostragem:** processo de selecionar os elementos que integrarão uma amostra (BARBETTA; BORNIA; REIS, 2010, seções 1.6 e 2.2).
- **Quadro amostral:** lista ou outra representação das unidades acessíveis à seleção.
- **Amostragem aleatória simples:** procedimento no qual cada amostra possível de mesmo tamanho tem a mesma probabilidade, isto é, a mesma chance, de ser selecionada.

Uma amostra obtida por esse sorteio é uma **amostra aleatória simples**. Nos exercícios, a seleção será **sem reposição**: um registro sorteado não pode ser escolhido novamente na mesma amostra.

Analisar todos os 344 registros descreve integralmente a população finita de referência. Isso não equivale a realizar um censo de todos os pinguins dessas espécies. Mesmo um censo pode conter erros de registro, medição ou omissão de unidades; observar toda a população não garante ausência de erros.

Nesta aula, as linhas disponíveis constituem o quadro amostral. O sorteio distribui as chances entre essas linhas; pinguins ausentes do arquivo não podem ser selecionados (NAVIDI, seção 1.1; PINHEIRO et al., seção 1.2).

### 6.1 Representatividade não é sinônimo de tamanho

A **cobertura** diz respeito a quais unidades da população de interesse estão incluídas no quadro amostral. A **representatividade** expressa a adequação da amostra para responder sobre essa população em relação às características estudadas; sua avaliação exige examinar a cobertura e o procedimento de seleção.

Uma amostra aleatória não precisa reproduzir exatamente a composição da população. Aumentar seu tamanho pode reduzir a variabilidade amostral, mas não corrige automaticamente a exclusão sistemática de grupos.

No Palmer Penguins, sortear mais linhas não inclui animais que ficaram fora da coleta original. Portanto, um sorteio dentro do arquivo não garante representatividade de uma população biológica mais ampla (BARBETTA; BORNIA; REIS, 2010, seção 2.2.1; NAVIDI, 2024, seção 1.1).

### 6.2 Variabilidade e viés não são a mesma coisa

Um **viés** é uma tendência sistemática de um procedimento a produzir resultados que se afastam da característica que se deseja conhecer. Nesta aula, distinguiremos problemas de seleção e de medição das diferenças ocasionais entre amostras.

| Fenômeno | Característica | Exemplo nesta aula |
| --- | --- | --- |
| Variabilidade amostral | Resultados mudam entre amostras obtidas pelo mesmo procedimento. | Duas amostras aleatórias de dez registros apresentam composições diferentes. |
| Viés de seleção | O procedimento favorece ou exclui sistematicamente unidades relevantes. | Filtrar somente Gentoo quando o objetivo é comparar as três espécies. |
| Viés de medição | O registro se afasta sistematicamente da característica que se pretende medir. | Deve ser considerado conceitualmente, mas não pode ser diagnosticado apenas com as saídas usadas nesta aula. |

A aleatoriedade aplicada depois de um filtro inadequado não recupera os grupos já excluídos. Por outro lado, diferenças ocasionais entre duas amostras aleatórias não bastam para concluir que houve viés.

### 6.3 Comparar amostras aleatórias reproduzíveis

Duas amostras de mesmo tamanho serão obtidas por amostragem aleatória simples sem reposição. No código, a **semente** é um valor que permite reproduzir a sequência usada pelo programa no sorteio; ela será indicada por `random_state`. Usaremos sementes diferentes para obter duas seleções.

A comparação permitirá examinar diferenças entre sorteios feitos pelo mesmo procedimento. A possibilidade de reproduzir um sorteio não garante a representatividade da amostra.

```python
# Sortear 10 observações sem reposição.
amostra_a = populacao_referencia.sample(
    n=10,
    random_state=42,
)

amostra_a
```

**Tabela 5 - Amostra aleatória A.** Foram sorteadas 10 linhas da população finita de referência. O valor de `random_state` permite reproduzir o mesmo resultado (PANDAS DEVELOPMENT TEAM, 2026). Fonte dos dados: Horst, Hill e Gorman (2020).

```python
# Repetir o sorteio com outra semente.
amostra_b = populacao_referencia.sample(
    n=10,
    random_state=7,
)

amostra_b
```

**Tabela 6 - Amostra aleatória B.** A mudança de `random_state` produz outra seleção reproduzível de 10 linhas. Fonte: elaboração própria com dados de Horst, Hill e Gorman (2020).

```python
# Contar as espécies presentes nas duas amostras.
composicao_amostra_a = amostra_a["Species"].value_counts()
composicao_amostra_b = amostra_b["Species"].value_counts()

# Reunir as contagens para facilitar a comparação.
comparacao_composicoes = pd.DataFrame(
    {
        "População de referência": composicao_referencia,
        "Amostra A": composicao_amostra_a,
        "Amostra B": composicao_amostra_b,
    }
).fillna(0).astype(int)

comparacao_composicoes
```

**Tabela 7 - Contagens por espécie na população e nas amostras.** As duas amostras foram obtidas pelo mesmo procedimento, mas apresentam contagens diferentes. Como a população possui 344 observações e cada amostra possui 10, a comparação proporcional será feita na próxima tabela. Fonte: elaboração própria com dados de Horst, Hill e Gorman (2020).


A **proporção** de uma espécie é sua quantidade de registros dividida pelo total de registros do conjunto analisado. Multiplicar essa razão por 100 a expressa em **porcentagem**. Usaremos 344 como total na população de referência e 10 em cada amostra; assim, poderemos comparar conjuntos de tamanhos diferentes.

```python
# Calcular as proporções de cada espécie, em porcentagem.
proporcoes_referencia = composicao_referencia / len(populacao_referencia) * 100
proporcoes_amostra_a = composicao_amostra_a / len(amostra_a) * 100
proporcoes_amostra_b = composicao_amostra_b / len(amostra_b) * 100

# Reunir e arredondar as porcentagens.
comparacao_proporcoes = pd.DataFrame(
    {
        "População de referência (%)": proporcoes_referencia,
        "Amostra A (%)": proporcoes_amostra_a,
        "Amostra B (%)": proporcoes_amostra_b,
    }
).fillna(0).round(1)

comparacao_proporcoes
```

**Tabela 8 - Proporções das espécies na população e nas amostras.** Adelie corresponde a 44,2% da população de referência, mas representa 10,0% da amostra A e 50,0% da amostra B. Gentoo passa de 36,0% na população para 60,0% na amostra A e 20,0% na amostra B. Essas diferenças ilustram a variabilidade da composição entre amostras aleatórias pequenas. Fonte: elaboração própria com dados de Horst, Hill e Gorman (2020).

**Registro do estudante:** a seleção aleatória produziu uma cópia exata da composição da população? Essas diferenças, isoladamente, comprovam viés?

**Resposta:** não. Nenhuma das amostras reproduziu exatamente a composição da população de referência. As diferenças observadas entre sorteios de dez registros são compatíveis com variabilidade amostral e, isoladamente, não comprovam viés. Para caracterizar viés, seria necessário identificar um mecanismo de seleção que favoreça ou exclua sistematicamente grupos relevantes ao objetivo.


```python
# Calcular a média da massa corporal em cada conjunto.
media_massa_referencia = populacao_referencia["Body Mass (g)"].mean()
media_massa_amostra_a = amostra_a["Body Mass (g)"].mean()
media_massa_amostra_b = amostra_b["Body Mass (g)"].mean()

# Reunir e arredondar as médias.
comparacao_medias = pd.Series(
    {
        "População de referência": media_massa_referencia,
        "Amostra A": media_massa_amostra_a,
        "Amostra B": media_massa_amostra_b,
    },
    name="Média da massa corporal (g)",
).round(1)

comparacao_medias
```

**Tabela 9 - Média da massa corporal na população e nas amostras.** A média é 4.201,8 g na população de referência, 4.262,5 g na amostra A e 3.872,5 g na amostra B. A mesma medida assume valores diferentes porque foi calculada em seleções diferentes. Isso evidencia variabilidade amostral e não demonstra, por si só, a existência de viés. Fonte: elaboração própria com dados de Horst, Hill e Gorman (2020).


### 6.4 O objetivo determina a adequação do filtro

Considere dois objetivos:

- **Objetivo A:** comparar Adelie, Chinstrap e Gentoo. Um filtro que mantém somente Gentoo é inadequado porque impede a comparação pretendida.
- **Objetivo B:** descrever somente os registros Gentoo. O filtro seleciona os registros da população de interesse definida pela pergunta.

O código não é adequado ou inadequado isoladamente. Sua validade depende da correspondência entre pergunta, população e procedimento de seleção.

#### Objetivo A: descrever as três espécies

### 7.1 Parâmetro, estatística amostral, estimador e estimativa

Um **parâmetro** $\theta$ descreve uma característica numérica da população. Uma **estatística amostral** é uma medida calculada na amostra, como sua média ou uma proporção (BARBETTA; BORNIA; REIS, 2010, seção 1.6).

Uma **variável aleatória** associa um valor numérico ao resultado de um processo sujeito ao acaso. Aqui, $X_i$ representa o valor que poderá ser observado na posição $i$ da amostra antes do sorteio; $x_i$ é o valor efetivamente observado e $n$ é o tamanho da amostra. A notação $T$ representa a regra de cálculo aplicada aos valores da amostra.

Um **estimador** é uma regra de cálculo utilizada para estimar um parâmetro a partir dos dados de uma amostra (PINHEIRO et al., 2009, seção 7.1):

$$
\widehat{\theta}=T(X_1,X_2,\ldots,X_n).
$$

A **estimativa** é o valor numérico obtido ao aplicar essa regra aos dados observados (PINHEIRO et al., 2009, seção 7.1):

$$
\hat{\theta}=T(x_1,x_2,\ldots,x_n).
$$

**Aplicação aos resultados anteriores:** a média de massa corporal dos registros preenchidos na população finita de referência é um parâmetro desse conjunto. A média calculada nos registros preenchidos de uma amostra é uma estatística amostral. Quando essa regra é usada para estimar a média da população, ela funciona como estimador; o número obtido é a estimativa. A interpretação se refere às massas disponíveis e exige considerar as ausências.

A correção do cálculo não garante a validade da generalização. Também são necessários uma população bem definida, um procedimento de seleção justificável e hipóteses compatíveis com o método inferencial (PINHEIRO et al., 2009, seção 7.1).

### 7.2 Como delimitar as conclusões

Uma conclusão responsável informa:

- quais observações foram analisadas;
- qual população foi definida;
- como as unidades foram selecionadas;
- quais grupos ou períodos ficaram fora da cobertura;
- se o resultado é descritivo, inferencial ou causal;
- quais limitações impedem uma generalização mais ampla.

No Palmer Penguins, uma formulação apropriada começa por “nos registros disponíveis” ou “na amostra selecionada”. A expressão “em todos os pinguins dessas espécies” exigiria justificativas que o arquivo, isoladamente, não fornece.

```python
# Calcular as contagens e as proporções observadas de cada espécie.
contagens_especies = populacao_referencia["Species"].value_counts()
proporcoes_especies = (contagens_especies / len(populacao_referencia) * 100).round(1)

# Calcular a quantidade válida, o menor e o maior valor de massa corporal.
massa_corporal = populacao_referencia["Body Mass (g)"]
amplitude_massa = pd.Series(
    {
        "Quantidade válida": massa_corporal.count(),
        "Mínimo": massa_corporal.min(),
        "Máximo": massa_corporal.max(),
    }
)

# Reunir as descrições em uma única tabela.
descricao_arquivo = pd.concat(
    [contagens_especies, proporcoes_especies, amplitude_massa],
    keys=["Contagem por espécie", "Proporção por espécie (%)", "Massa corporal (g)"],
    names=["Medida", "Categoria"],
).to_frame(name="Valor")

descricao_arquivo
```

**Tabela 13 - Descrição do arquivo por espécie e massa corporal.** As contagens, as proporções e os valores mínimo e máximo pertencem às observações disponíveis. Fonte: elaboração própria com dados de Horst, Hill e Gorman (2020).

**Registro do estudante:** escreva uma conclusão descritiva sustentada pela saída e uma generalização que a seleção disponível não permite sustentar.

**Resposta:** no arquivo, Adelie possui 152 registros (44,2%), Gentoo 124 (36,0%) e Chinstrap 68 (19,8%). Entre os 342 valores não ausentes de massa corporal, os valores variam de 2.700 g a 6.300 g. Esses resultados não permitem afirmar que as mesmas proporções ou os mesmos valores mínimo e máximo caracterizam todos os pinguins dessas espécies, pois o arquivo não é uma amostra probabilística dessa população biológica.
