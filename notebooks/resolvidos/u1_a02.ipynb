# Aula 2 — Fundamentos estatísticos e investigação com dados

## Carregamento do conjunto de dados

```python
# Carregar e selecionar as variáveis usadas na aula.
import pandas as pd

url = "https://raw.githubusercontent.com/carubbi/MQ/main/data/raw/penguins_raw.csv"
dados = pd.read_csv(url)
colunas = [
    "studyName",
    "Individual ID",
    "Species",
    "Island",
    "Body Mass (g)",
]
dados = dados[colunas]
```

```python
# Consultar a estrutura do conjunto de dados.
dados.info()
```

## Ciclo 1 — reconhecer a variabilidade

```python
# Contar os valores válidos de massa corporal.
massa = dados["Body Mass (g)"]
massa.count()
```

```python
# Identificar a menor massa corporal.
massa.min()
```

```python
# Identificar a maior massa corporal.
massa.max()
```

```python
# Contar os valores distintos de massa corporal.
massa.nunique()
```

## Ciclo 2 — comparar descrição amostral e populacional

```python
# Selecionar uma amostra aleatória com massas válidas.
validos = dados.dropna(subset=["Body Mass (g)"])
amostra = validos.sample(n=30, random_state=42)
```

```python
# Calcular a média da massa corporal no conjunto disponível.
media = massa.mean()
media
```

```python
# Calcular a média da massa corporal na amostra.
massa_amostra = amostra["Body Mass (g)"]
media_amostra = massa_amostra.mean()
media_amostra
```

## Ciclo 4 — comparar amostragem aleatória e por conveniência

```python
# Selecionar uma amostra aleatória da população de referência.
amostra_aleatoria = dados.sample(n=30, random_state=42)
```

```python
# Selecionar uma amostra por conveniência na ilha Biscoe.
ilha = dados["Island"]
filtro = ilha == "Biscoe"
biscoe = dados[filtro]
amostra_conveniencia = biscoe.head(30)
```

```python
# Contar as espécies na amostra aleatória.
especies = amostra_aleatoria["Species"]
contagem_aleatoria = especies.value_counts()
contagem_aleatoria
```

```python
# Contar as espécies na amostra por conveniência.
especies = amostra_conveniencia["Species"]
contagem_conveniencia = especies.value_counts()
contagem_conveniencia
```
