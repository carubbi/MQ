## 3. Experimento, espaço amostral e eventos

Como calcular a probabilidade de um evento?


### 3.1 Experimento aleatório

Um **experimento aleatório** é um procedimento cujo resultado particular é incerto antes de sua realização. Para descrevê-lo, indicamos o procedimento e o que será registrado como resultado. Ao lançar uma moeda, por exemplo, registramos a face voltada para cima (Barbetta; Reis; Bornia, 2010, pp. 92, 96).


### 3.2 Espaço amostral

O **espaço amostral** $\Omega$ reúne todos os resultados possíveis do experimento. No lançamento da moeda, $\Omega_{\text{moeda}}=\{\text{cara},\text{coroa}\}$. Se lançamos um dado e registramos a face voltada para cima, $\Omega_{\text{dado}}=\{1,2,3,4,5,6\}$. Especificar o que registramos determina quais resultados entram no espaço amostral (Barbetta; Reis; Bornia, 2010, p. 93).


### 3.3 Eventos

Um **evento** é um subconjunto do espaço amostral. Ele ocorre quando o resultado observado pertence a esse subconjunto. Na moeda, o evento “sair cara” é $C=\{\text{cara}\}$. No dado, “sair número par” é $D=\{2,4,6\}$. Se a face observada for 4, ocorreu $D$. Esses conjuntos descrevem possibilidades. Atribuir a mesma probabilidade às faces do dado exige que ele seja honesto e o lançamento, imparcial (Barbetta; Reis; Bornia, 2010, pp. 92, 94, 96).


### 3.4 Axiomas e cálculo da probabilidade

**Probabilidade** é uma função $\mathbb{P}$ que atribui a cada evento $E$ do espaço amostral $\Omega$ um número $\mathbb{P}(E)$ entre 0 e 1 (Montgomery; Runger, 2018, seção 2.3, p. 47). Na moeda honesta com lançamento imparcial, as duas faces têm a mesma probabilidade e $\mathbb{P}(C)=1/2$. Essa igualdade depende da hipótese de equiprobabilidade (Barbetta; Reis; Bornia, 2010, pp. 96–97).

Quando todos os resultados possíveis têm a mesma probabilidade, calculamos a probabilidade pela razão entre resultados favoráveis e resultados possíveis (Barbetta; Reis; Bornia, 2010, pp. 96–97):

$$\mathbb{P}(E)=\frac{n(E)}{n(\Omega)}.
\tag{1.1}
$$

- $n(E)$: número de resultados que pertencem ao evento $E$.
- $n(\Omega)$: número total de resultados possíveis.

No espaço amostral finito desta aula, uma atribuição de probabilidades $\mathbb{P}$ deve satisfazer três axiomas para quaisquer eventos $E$ e $F$ (Montgomery; Runger, 2018, seção 2.3, p. 47):

1. **Não negatividade:** $\mathbb{P}(E)\geq 0$.
2. **Normalização:** $\mathbb{P}(\Omega)=1$.
3. **Aditividade para eventos disjuntos:** se $E\cap F=\varnothing$, então $\mathbb{P}(E\cup F)=\mathbb{P}(E)+\mathbb{P}(F)$.

Em um espaço amostral finito, calculamos a probabilidade de um evento somando as probabilidades de seus resultados elementares (Morettin; Bussab, 2010, p. 122):

$$\mathbb{P}(E)=\sum_{\omega\in E}\mathbb{P}(\{\omega\}).
\tag{1.2}
$$

- $\mathbb{P}(E)$: probabilidade do evento $E$.
- $\omega$: um resultado possível do espaço amostral finito.
- $\mathbb{P}(\{\omega\})$: probabilidade desse resultado.

A função `prob` reúne a contagem de `True` e a divisão pelo total. No sorteio uniforme entre os 1.000 CDs, ela calcula a probabilidade no lote. Em uma amostra simulada, calcula a frequência relativa. Ela não atribui pesos diferentes às linhas: aplicá-la às quatro categorias como se fossem equiprováveis seria incorreto. Um evento sempre falso tem probabilidade zero; uma série vazia não fornece um denominador válido.

```python
# Validar a representação booleana de um evento
def validar_evento(A):
    """Verifica a representação de um evento por uma série booleana."""
    if not isinstance(A, pd.Series):
        raise TypeError("O evento deve ser uma série do pandas.")
    if not pd.api.types.is_bool_dtype(A.dtype):
        raise TypeError("A série deve conter valores booleanos.")
    if A.isna().any():
        raise ValueError("Trate os valores ausentes antes do cálculo.")

# Calcular a proporção de registros que pertencem ao evento
def prob(A):
    """Calcula a proporção de True, atribuindo peso igual às linhas."""
    validar_evento(A)
    if A.empty:
        raise ValueError("O conjunto de referência está vazio.")
    return A.sum() / A.size
```

## 4. Complemento, união e interseção


### 4.1 Complemento

O **complemento** $A^{c}$ contém os resultados de $\Omega$ que não estão em $A$ (Barbetta; Reis; Bornia, 2010, pp. 94–96).

$$\mathbb{P}(A^{c})=1-\mathbb{P}(A).
\tag{1.3}
$$

- $A^{c}$: evento em que $A$ não ocorre.
- $1$: probabilidade do evento certo $\Omega$.

O evento impossível $\varnothing$ tem probabilidade zero.


### 4.2 União e interseção

A **união** $A\cup B$ contém os resultados que estão em pelo menos um dos eventos. A **interseção** $A\cap B$ contém apenas os que estão em ambos. “Ou” é inclusivo: um resultado na interseção também pertence à união (Barbetta; Reis; Bornia, 2010, pp. 94–102; Escovedo; Marques; Kalinowski, 2025, pp. 82–86; Morettin; Bussab, 2010, pp. 123–125).

## 5. Regra da soma e exclusão mútua


### 5.1 Regra geral da soma

Ao somar $\mathbb{P}(A)$ e $\mathbb{P}(B)$, os resultados de $A\cap B$ são contados duas vezes. A **regra geral da soma** corrige essa sobreposição (Barbetta; Reis; Bornia, 2010, pp. 97–102; Morettin; Bussab, 2010, pp. 123–125):

$$\mathbb{P}(A\cup B)=\mathbb{P}(A)+\mathbb{P}(B)-\mathbb{P}(A\cap B).
\tag{1.4}
$$

- $\mathbb{P}(A\cup B)$: probabilidade de ao menos um dos eventos.
- $\mathbb{P}(A\cap B)$: probabilidade da ocorrência simultânea.

## 6. Probabilidade condicional e regra do produto

Até aqui, calculamos as probabilidades em relação aos 1.000 CDs do lote. Agora veremos como a informação de que um evento ocorreu muda o conjunto de referência. Depois, usaremos uma probabilidade condicional para calcular a ocorrência conjunta.

### 6.1 Condicionamento

Ao saber que $B$ ocorreu, restringimos a referência aos resultados de $B$. A probabilidade de $A$ **dado** $B$ é a razão entre a probabilidade da interseção e a de $B$ (Barbetta; Reis; Bornia, 2010, pp. 103–105; Morettin; Bussab, 2010, p. 128):

$$\mathbb{P}(A\mid B)=\frac{\mathbb{P}(A\cap B)}{\mathbb{P}(B)},\qquad \mathbb{P}(B)>0.
\tag{1.6}
$$

- $\mathbb{P}(A\mid B)$: probabilidade de resistência alta entre os CDs com trilhas aprovadas.
- $\mathbb{P}(B)$: probabilidade do evento condicionante, que precisa ser positiva.

Em `prob_cond(A, B)`, a expressão `A[B]` seleciona os valores de A somente nos registros que satisfazem B. A função `prob` passa a dividir pelo tamanho desse subconjunto. As duas séries devem representar os mesmos indivíduos. Se nenhum registro satisfizer B, a condicional não pode ser calculada nesse conjunto.

```python
# Restringir o conjunto de referência ao evento condicionante
def prob_cond(A, B):
    """Calcula a proporção de A entre os registros que satisfazem B."""
    validar_evento(A)
    validar_evento(B)
    if not A.index.equals(B.index):
        raise ValueError("A e B devem ter os mesmos índices e ordem.")
    if not B.any():
        raise ValueError("Não há registros que satisfaçam B.")
    return prob(A[B])
```

$$

$$\mathbb{P}(A\cap B)=\mathbb{P}(B)\mathbb{P}(A\mid B).

$$

## 7. Partições e probabilidade total

Como atualizar uma probabilidade após observar uma nova evidência?

### 7.1 Partição do espaço amostral

Uma **partição** divide o espaço amostral em grupos que não se sobrepõem e que, juntos, cobrem todos os resultados possíveis. Por exemplo, ao lançar um dado, “face par” e “face ímpar” formam uma partição. Cada resultado pertence a exatamente um dos grupos (Barbetta; Reis; Bornia, 2010, pp. 110–111).

$$
G_i\cap G_j=\varnothing\quad(i\ne j),\qquad \bigcup_{i=1}^{k}G_i=\Omega.\tag{1.8}
$$

Aqui, $\Omega$ é o espaço amostral, $G_1,\ldots,G_k$ são os grupos da partição, $\cap$ representa a interseção, $\cup$ representa a união e $\varnothing$ é o conjunto vazio. Para usar uma probabilidade condicionada a $G_i$, exigimos $\mathbb{P}(G_i)>0$.

### 7.2 Regra da probabilidade total

Se a evidência $E$ pode ocorrer em qualquer grupo da partição, podemos separá-la em parcelas disjuntas (Barbetta; Reis; Bornia, 2010, pp. 110–111).

$$
E=\bigcup_{i=1}^{k}(E\cap G_i).\tag{1.9}
$$

Pela regra do produto, a probabilidade de cada parcela é a probabilidade inicial do grupo multiplicada pela probabilidade de $E$ nesse grupo. A soma das parcelas fornece a **probabilidade total** de $E$ (Barbetta; Reis; Bornia, 2010, p. 111).

$$
\mathbb{P}(E)=\sum_{i=1}^{k}\mathbb{P}(G_i)\mathbb{P}(E\mid G_i).\tag{1.10}
$$

Aqui, $\mathbb{P}(G_i)$ é a probabilidade inicial do grupo, $\mathbb{P}(E\mid G_i)$ é a probabilidade da evidência dentro dele e $\mathbb{P}(E)$ é a probabilidade da evidência considerando todos os grupos. Não basta somar as probabilidades condicionais. Cada uma deve receber o peso do seu grupo.

## 8. Teorema de Bayes e probabilidade inicial

### 8.1 Atualização da probabilidade

Seja $G$ o **grupo de interesse** e $E$ a **evidência observada**, com $\mathbb{P}(G)>0$ e $\mathbb{P}(E)>0$. A probabilidade **a priori** é $\mathbb{P}(G)$. Após observar $E$, a probabilidade **a posteriori** é $\mathbb{P}(G\mid E)$ (MORETTIN; BUSSAB, 2010, pp. 116–118).

Pela definição de probabilidade condicional e pela regra do produto (BARBETTA; REIS; BORNIA, 2010, pp. 105, 112):

$$
\begin{aligned}
\mathbb{P}(G\mid E)&=\frac{\mathbb{P}(G\cap E)}{\mathbb{P}(E)},\\
\mathbb{P}(G\cap E)&=\mathbb{P}(G)\mathbb{P}(E\mid G).
\end{aligned}
$$

Substituindo a segunda identidade na primeira:

$$
\mathbb{P}(G\mid E)=\frac{\mathbb{P}(G)\mathbb{P}(E\mid G)}{\mathbb{P}(E)}.\tag{1.11}
$$

Pela probabilidade total da Equação (1.10):

$$
\mathbb{P}(G\mid E)=\frac{\mathbb{P}(G)\mathbb{P}(E\mid G)}{\sum_{i=1}^{k}\mathbb{P}(G_i)\mathbb{P}(E\mid G_i)}.\tag{1.12}
$$

## 9. Independência


### 9.1 Verificação da independência

Dois eventos são **independentes** quando saber que um ocorreu não altera a probabilidade do outro. Para dois eventos, podemos verificar a seguinte condição (Barbetta; Reis; Bornia, 2010, pp. 109–110; Morettin; Bussab, 2010, pp. 129–131):

$$\mathbb{P}(A\cap B)=\mathbb{P}(A)\mathbb{P}(B).
\tag{1.15}
$$

Esta fatoração é uma condição a verificar, não um atalho automático. Eventos mutuamente exclusivos de probabilidades positivas não são independentes: a ocorrência de um impede a do outro.

Na tabela a seguir, os números **já são probabilidades** (Morettin; Bussab, 2010, problema 21, p. 132):
