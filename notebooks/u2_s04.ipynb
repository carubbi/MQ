### 3.2. Fundamentação teórica

O espaço amostral $\Omega$ reúne os resultados possíveis de um experimento aleatório. $\omega\in\Omega$ representa um desses resultados. Uma variável aleatória real $X:\Omega\to\mathbb{R}$ associa um número real a cada resultado. O símbolo $X$ representa a variável. $x$ representa um valor que ela pode assumir.

Considere uma variável aleatória contínua que admite densidade $f_X$. As condições de uma função densidade são

$$
f_X(x)\geq0,\qquad \int_{-\infty}^{\infty} f_X(x)\,dx=1.
\tag{7.1}
$$

Em $f_X(x)$, o índice $X$ identifica a variável aleatória, e $x$ indica o valor em que calculamos a densidade. No exemplo do ponteiro, $X$ representa o ângulo final, e $f_X(45)$ representa a densidade em 45 graus.

$f_X(x)$ é a densidade no valor $x$. A integral representa a área total sob a curva. A densidade não precisa ser constante.

O suporte indica onde a distribuição concentra sua probabilidade. A probabilidade de $X$ assumir valores fora do suporte é zero.

A função de distribuição acumulada é

$$
F_X(x)=\mathbb{P}(X\leq x)=\int_{-\infty}^{x}f_X(v)\,dv.
\tag{7.2}
$$

$F_X(x)$ é a probabilidade acumulada até $x$. $v$ é a variável de integração. Se $X$ tem uma unidade de medida, a densidade tem a unidade inversa. No exemplo do ponteiro, $X$ é medido em graus e a densidade $f_X(x)$, em $\mathrm{grau}^{-1}$. A probabilidade e a função de distribuição $F_X(x)$ não têm unidade.

Para limites reais $c<d$,

$$
\mathbb{P}(c<X\leq d)=F_X(d)-F_X(c),\qquad \mathbb{P}(X=x)=0.
\tag{7.3}
$$

$c$ e $d$ delimitam o intervalo de interesse. A inclusão dos extremos não altera a probabilidade, pois um ponto tem área nula. Isso não impede que uma medição arredondada apresente valores repetidos (Barbetta; Reis; Bornia, 2010, p. 141–145; Pinheiro *et al.*, 2009, p. 111–112).

Quando $\mathbb{E}[X^2]<\infty$, a esperança e a variância são finitas e dadas por

$$
\mathbb{E}[X]=\int_{-\infty}^{\infty}x f_X(x)\,dx,
\qquad
\operatorname{Var}(X)=\mathbb{E}\!\left[(X-\mathbb{E}[X])^2\right].
\tag{7.4}
$$

$\mathbb{E}[X^2]$ é o valor esperado do quadrado de $X$, chamado segundo momento. $\mathbb{E}[X]$ descreve o centro da distribuição, na unidade de $X$. $\operatorname{Var}(X)$ mede a dispersão em torno da esperança, na unidade ao quadrado (Barbetta; Reis; Bornia, 2010, p. 145–146).

### 4.2. Fundamentação teórica

Uma variável aleatória $U:\Omega\to[a,b]$ tem distribuição Uniforme contínua com parâmetros reais $a<b$ quando sua densidade é constante nesse intervalo e nula fora dele. Escrevemos $U\sim\operatorname{Uniforme}(a,b)$. $u$ representa um valor da variável. Subintervalos do suporte com a mesma amplitude têm a mesma probabilidade.

$$
f_U(u)=\begin{cases}
\dfrac{1}{b-a},&a\leq u\leq b,\\
0,&\text{caso contrário}.
\end{cases}
\tag{7.5}
$$

$f_U(u)$ é a densidade. $a$ e $b$ são os limites inferior e superior do suporte. $b-a$ é sua amplitude. Os parâmetros têm a mesma unidade de $U$, e a densidade tem a unidade inversa.

$$
F_U(u)=\begin{cases}
0,&u<a,\\
\dfrac{u-a}{b-a},&a\leq u<b,\\
1,&u\geq b.
\end{cases}
\tag{7.6}
$$

$F_U(u)$ é a probabilidade acumulada até $u$. Para limites reais $c<d$, a probabilidade do intervalo é $F_U(d)-F_U(c)$. A cauda direita em $u$ é $1-F_U(u)$. A definição por partes também trata limites fora do suporte.

$$
\mathbb{E}[U]=\frac{a+b}{2},\qquad
\operatorname{Var}(U)=\frac{(b-a)^2}{12}.
\tag{7.7}
$$

A esperança é o ponto médio do suporte, na unidade de $U$. A variância depende do quadrado da amplitude e tem a unidade de $U$ ao quadrado (Barbetta; Reis; Bornia, 2010, p. 148; Morettin; Bussab, 2010, p. 174–175).

## 5. Distribuição Exponencial

### 5.1. Contexto

O tempo de vida de um transistor segue uma Exponencial com média de 500 horas. **Qual é a probabilidade de durar mais de 500 horas? E entre 300 e 1000 horas?** (Barbetta; Reis; Bornia, 2010, p. 152, exercício 6, itens a–b).

O modelo admite taxa de falha constante, sem envelhecimento. Essa hipótese precisa ser avaliada em uma aplicação real: conhecer apenas o tempo médio não justifica uma Exponencial.

### 5.2. Fundamentação teórica

Uma variável aleatória não negativa $T:\Omega\to[0,\infty)$ tem distribuição Exponencial com parâmetro de taxa $\lambda>0$ quando sua densidade tem a forma abaixo. Escrevemos $T\sim\operatorname{Exp}(\lambda)$. $t$ representa um valor da variável. O parâmetro $\lambda$ tem unidade inversa à de $T$.

$$
f_T(t)=\begin{cases}\lambda e^{-\lambda t},&t\geq0,\\0,&t<0.\end{cases}
\tag{7.8}
$$

$f_T(t)$ é a densidade em $t$ e $e$ é a base da função exponencial. O produto $\lambda t$ não tem unidade.

$$
F_T(t)=\begin{cases}1-e^{-\lambda t},&t\geq0,\\0,&t<0.\end{cases}
\tag{7.9}
$$

$F_T(t)$ é a probabilidade de $T$ assumir um valor menor ou igual a $t$. Sua complementar é a **função de sobrevivência**:

$$
S_T(t)=\mathbb{P}(T>t)=\begin{cases}e^{-\lambda t},&t\geq0,\\1,&t<0.\end{cases}
\tag{7.10}
$$

$S_T(t)$ é a probabilidade de $T$ exceder $t$. Para limites reais $c<d$, a probabilidade do intervalo é $F_T(d)-F_T(c)=S_T(c)-S_T(d)$. As definições por partes também tratam limites negativos.

$$
\mathbb{E}[T]=\frac{1}{\lambda},\qquad \operatorname{Var}(T)=\frac{1}{\lambda^2}.
\tag{7.11}
$$

A esperança tem a unidade de $T$. A variância tem essa unidade ao quadrado. A **escala** é $\beta=1/\lambda$, na mesma unidade de $T$. Essa é a parametrização utilizada por Morettin e Bussab: taxa e escala são recíprocas (Barbetta; Reis; Bornia, 2010, p. 149–152; Morettin; Bussab, 2010, p. 181–182; Pinheiro *et al.*, 2009, p. 117–118).

Em outro contexto, retomaremos o transistor: **se já funcionou 500 horas, qual é a probabilidade de funcionar por mais 500 horas?** O mecanismo de chegadas do servidor e a hipótese de vida útil do transistor devem ser examinados separadamente.

### 6.2. Fundamentação teórica

Um processo de contagem associa a cada instante $t\geq0$ uma variável aleatória $N(t):\Omega\to\{0,1,2,\ldots\}$. O espaço amostral $\Omega$ reúne as trajetórias possíveis do processo, e $N(t)$ conta as ocorrências em $(0,t]$. A variável $W:\Omega\to[0,\infty)$ representa o tempo até a primeira ocorrência, a partir do instante zero.

Em um **processo de Poisson homogêneo** de taxa $\lambda>0$, $N(0)=0$, as contagens em intervalos disjuntos são independentes e a distribuição da contagem depende apenas da duração do intervalo. A contagem em um intervalo de duração $t$ tem distribuição Poisson de média $\lambda t$:

$$
m_t=\lambda t,\qquad N(t)\sim\operatorname{Poisson}(m_t),\qquad t\geq0.
\tag{7.12}
$$

$\lambda$ é a taxa de ocorrências por unidade de tempo. $t$ é a duração do intervalo. $m_t=\lambda t$ é a média da contagem nesse intervalo. Taxa e média da contagem são parâmetros distintos. Uma distribuição marginal Poisson, isoladamente, não estabelece os pressupostos do processo.

$$
\mathbb{P}(W>t)=\mathbb{P}(N(t)=0)=e^{-m_t}=e^{-\lambda t}.
\tag{7.13}
$$

Os eventos $W>t$ e $N(t)=0$ são equivalentes: esperar mais que $t$ pela primeira ocorrência significa não observar ocorrências até $t$. Assim, $W\sim\operatorname{Exp}(\lambda)$. A taxa do processo determina a distribuição da espera (Barbetta; Reis; Bornia, 2010, p. 149–150; Pinheiro *et al.*, 2009, p. 117–118).

Para qualquer variável $T\sim\operatorname{Exp}(\lambda)$, a **propriedade de falta de memória** é

$$
\mathbb{P}(T>s+t\mid T>s)
=\frac{S_T(s+t)}{S_T(s)}
=\frac{e^{-\lambda(s+t)}}{e^{-\lambda s}}
=e^{-\lambda t}
=\mathbb{P}(T>t),\qquad s,t\geq0.
\tag{7.14}
$$

$s$ representa um valor já ultrapassado. $t$ é o acréscimo considerado. $s+t$ é o limite total. Se a variável representa uma duração, esses valores correspondem ao tempo decorrido, ao tempo adicional e ao tempo total. A razão decorre de $\{T>s+t\}\subseteq\{T>s\}$, com $S_T(s)>0$. A igualdade é uma propriedade da distribuição Exponencial (Barbetta; Reis; Bornia, 2010, p. 152, exercícios 6–7).

### 7.2. Fundamentação teórica

Uma variável aleatória real $X:\Omega\to\mathbb{R}$ tem distribuição Normal com parâmetros $\mu\in\mathbb{R}$ e $\sigma>0$ quando sua densidade tem a forma abaixo. Escrevemos $X\sim\mathcal{N}(\mu,\sigma^2)$. $x$ representa um valor da variável. Os parâmetros $\mu$ e $\sigma$ representam, respectivamente, a média e o desvio-padrão.

$$
f_X(x)=\frac{1}{\sigma\sqrt{2\pi}}\exp\!\left[-\frac{(x-\mu)^2}{2\sigma^2}\right],\qquad x\in\mathbb{R}.
\tag{8.1}
$$

$f_X(x)$ é a densidade em $x$. $\pi$ é a constante pi e $\exp$ é a função exponencial. $\mu$, $\sigma$ e $x$ têm a mesma unidade, e a densidade tem a unidade inversa. A curva é simétrica em torno de $\mu$. $\sigma$ controla sua dispersão.

$$
\mathbb{E}[X]=\mu,\qquad \operatorname{Var}(X)=\sigma^2.
\tag{8.2}
$$

$\sigma^2$ é a variância, na unidade de $X$ ao quadrado. Na notação $\mathcal{N}(\mu,\sigma^2)$, o segundo parâmetro é a variância.

O suporte da Normal é $\mathbb{R}$: sua densidade é positiva para todo valor real. O parâmetro $\mu$ determina a localização, mas não restringe o suporte (Barbetta; Reis; Bornia, 2010, p. 152–155; Morettin; Bussab, 2010, p. 176–178).

### 8.2. Fundamentação teórica

Considere $X\sim\mathcal{N}(\mu,\sigma^2)$, com $\mu\in\mathbb{R}$ e $\sigma>0$. A variável padronizada e o escore correspondente a um valor $x$ são

$$
Z=\frac{X-\mu}{\sigma}\sim\mathcal{N}(0,1),\qquad
z=\frac{x-\mu}{\sigma},\qquad \sigma>0.
\tag{8.3}
$$

$Z:\Omega\to\mathbb{R}$ é definida sobre o mesmo espaço amostral de $X$. $z$ é o valor de $Z$ correspondente a $X=x$. $\mu$ é a média e $\sigma$ é o desvio-padrão de $X$. O quociente não tem unidade. Um escore positivo indica um valor acima da média. Um escore negativo indica um valor abaixo da média. Seu módulo mede a distância em desvios-padrão.

$$
\mathbb{P}(X\leq x)=\mathbb{P}\!\left(Z\leq\frac{x-\mu}{\sigma}\right).
\tag{8.4}
$$

A transformação é crescente, pois $\sigma>0$, e preserva a probabilidade do evento correspondente. $Z$ tem distribuição Normal padrão, com média zero e variância um, **porque $X$ é Normal**. Padronizar uma variável de outra distribuição, com média e variância finitas e variância positiva, produz média zero e variância um, mas não a torna Normal (Barbetta; Reis; Bornia, 2010, p. 155–156; Morettin; Bussab, 2010, p. 177–178).

## 9. Cálculo de probabilidades na distribuição Normal

### 9.1. Contexto

### 9.2. Fundamentação teórica

Para uma variável $Z\sim\mathcal{N}(0,1)$, a densidade e a função de distribuição acumulada são

$$
\phi(z)=\frac{1}{\sqrt{2\pi}}e^{-z^2/2},\qquad
\Phi(z)=\int_{-\infty}^{z}\phi(v)\,dv.
\tag{8.5}
$$

A função $\phi$ representa a densidade. A função $\Phi$ representa a probabilidade acumulada. O símbolo $z$ é um valor real na escala padronizada, e $v$ é a variável de integração. A altura de $\Phi$ já é uma probabilidade: não corresponde a uma área sob a curva da acumulada.

Para $X\sim\mathcal{N}(\mu,\sigma^2)$, com $\mu\in\mathbb{R}$ e $\sigma>0$,

$$
\begin{aligned}
F_X(x)&=\Phi\!\left(\frac{x-\mu}{\sigma}\right),\\
\mathbb{P}(X>x)&=1-F_X(x),\\
\mathbb{P}(c<X\leq d)&=F_X(d)-F_X(c),\qquad c<d.
\end{aligned}
\tag{8.6}
$$

$F_X$ é a acumulada de $X$. $\mu$ e $\sigma$ são sua média e seu desvio-padrão. $c<d$ são limites reais na mesma escala de $X$. A primeira expressão relaciona as acumuladas nas escalas original e padronizada. As demais fornecem probabilidades de cauda e de intervalo. A inclusão ou exclusão dos extremos não altera essas probabilidades.

Pela simetria da Normal padrão,

$$
\Phi(-z)=1-\Phi(z).
\tag{8.7}
$$

Essa identidade vale para qualquer $z\in\mathbb{R}$ e relaciona as probabilidades acumuladas em escores opostos (Barbetta; Reis; Bornia, 2010, p. 154–158; Morettin; Bussab, 2010, p. 177–178).

```python
# ÁREAS DOS EVENTOS ORIGINAIS
fig, eixos = plt.subplots(1, 2, figsize=(12, 4))
for ax in eixos:
    ax.plot(x_n, modelo_x.pdf(x_n))
    ax.set(xlabel='Tempo de resposta (s)', ylabel='Densidade (s⁻¹)')
x_esquerda = np.linspace(7, 25, 361)
eixos[0].fill_between(x_esquerda, modelo_x.pdf(x_esquerda), alpha=0.3)
eixos[0].axvline(25, color='black', linestyle=':')
eixos[0].set_title(f'Item (a): P(X < 25) = {prob_x_a:.4f}')
eixos[1].fill_between(x_evento, modelo_x.pdf(x_evento), alpha=0.3)
eixos[1].axvline(20, color='black', linestyle=':')
eixos[1].axvline(30, color='black', linestyle=':')
eixos[1].set_title(f'Item (b): P(20 < X < 30) = {prob_x_b:.4f}')
fig.tight_layout()
plt.show()
```
