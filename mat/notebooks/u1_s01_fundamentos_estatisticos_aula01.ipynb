## 3. Diferenciar `DataFrame` e `Series`

Um `DataFrame` representa uma tabela bidimensional. Uma coluna selecionada isoladamente é uma `Series`, estrutura unidimensional que preserva os rótulos das linhas, o nome e o tipo computacional.

````python
# Selecionar uma coluna como Series.
especies = df["Species"]

especies.head()
````

## 4. Selecionar uma ou várias colunas

Para selecionar várias colunas, usamos uma lista de nomes. Os colchetes internos formam a lista; os externos realizam a seleção no `DataFrame`.

````python
colunas_anatomicas = [
    "Culmen Length (mm)",
    "Culmen Depth (mm)",
    "Flipper Length (mm)",
]

medidas = df[colunas_anatomicas]

medidas.head()
````

Como a seleção mantém várias colunas, `medidas` continua sendo um `DataFrame`. Essa diferença entre uma coluna e uma lista de colunas é importante para prever o tipo do resultado.

## 5. Selecionar por posição com `iloc`

`iloc` seleciona linhas e colunas por **posição inteira**:

```text
df.iloc[linhas, colunas]
```

O trecho antes da vírgula controla as linhas; o trecho depois controla as colunas. O símbolo `:` significa todas as posições daquele eixo.

````python
# Selecionar as cinco primeiras posições de linha e todas as colunas.
primeiras_linhas = df.iloc[0:5, :]

primeiras_linhas
````

Em um intervalo posicional, o início é incluído e o limite final é excluído. Por isso, `0:5` seleciona as posições `0`, `1`, `2`, `3` e `4`.

````python
# Selecionar cinco linhas e as colunas das posições 2, 3 e 4.
recorte_posicional = df.iloc[0:5, 2:5]

recorte_posicional
````

O segundo intervalo, `2:5`, seleciona as colunas das posições 2 a 4. A posição 5 não entra no resultado.

## 6. Selecionar por rótulos com `loc`

`loc` usa os **rótulos das linhas** e os **nomes das colunas**. No índice atual, os rótulos das primeiras linhas são números inteiros, mas continuam sendo rótulos, não posições.

````python
colunas_identificacao = [
    "Individual ID",
    "Species",
    "Island",
]

recorte_rotulado = df.loc[0:4, colunas_identificacao]

recorte_rotulado
````

Em `loc`, um intervalo de rótulos inclui o rótulo final quando ele existe. Portanto, `0:4` retorna os rótulos `0`, `1`, `2`, `3` e `4`.

## 7. Filtrar linhas por uma condição

Uma comparação aplicada a uma coluna produz uma `Series` booleana: cada observação recebe `True` quando satisfaz a condição e `False` quando não satisfaz.

````python
# Verificar quais observações pertencem à ilha Dream.
condicao = df["Island"] == "Dream"

condicao.head()
````

A condição possui o mesmo índice do `DataFrame`. Essa correspondência permite usá-la para selecionar somente as linhas marcadas como `True`.

````python
colunas_resultado = [
    "Species",
    "Island",
    "Body Mass (g)",
]

pinguins_dream = df.loc[condicao, colunas_resultado]

pinguins_dream.head()
````

`loc` combina duas decisões: antes da vírgula, mantém somente os pinguins da ilha Dream; depois da vírgula, mantém somente as três variáveis escolhidas.

## 8. Criar uma coluna derivada

Para preservar os dados carregados, primeiro criaremos uma cópia de trabalho. Alterações realizadas em `dados` não serão aplicadas a `df`.

````python
# Criar uma cópia independente para as transformações.
dados = df.copy()
````

````python
# Converter a massa corporal de gramas para quilogramas.
dados["Body Mass (kg)"] = dados["Body Mass (g)"] / 1000
````

A divisão é aplicada elemento a elemento à coluna inteira. Não é necessário escrever um laço para percorrer as linhas.

````python
colunas_massa = [
    "Body Mass (g)",
    "Body Mass (kg)",
]

comparacao_massa = dados.loc[:, colunas_massa]

comparacao_massa.head()
````

A nova coluna aparece somente em `dados`. Manter o objeto original preservado facilita conferir transformações e reiniciar a análise quando necessário.

## 9. Acrescentar uma linha a uma tabela

Inserir uma linha fictícia diretamente no conjunto científico misturaria observações reais e simuladas. Por isso, a operação será demonstrada em uma tabela pequena chamada `cadastro_exemplo`.

Primeiro, selecionaremos as colunas da demonstração.

````python
colunas_cadastro = [
    "Individual ID",
    "Species",
    "Island",
]

cadastro_exemplo = df.loc[:, colunas_cadastro]
````

Agora usaremos `iloc` para manter somente as duas primeiras posições de linha.

````python
cadastro_exemplo = cadastro_exemplo.iloc[0:2, :]

cadastro_exemplo
````

````python
# Tornar o recorte independente antes de modificá-lo.
cadastro_exemplo = cadastro_exemplo.copy()
````

Um novo registro pode ser representado por um `DataFrame` de uma linha. Os valores deixam explícito que se trata de uma simulação didática.

````python
novo_registro = pd.DataFrame(
    [{
        "Individual ID": "EXEMPLO",
        "Species": "registro fictício",
        "Island": "não se aplica",
    }]
)

novo_registro
````

`pd.concat()` combina tabelas. Com `ignore_index=True`, o resultado recebe um novo índice sequencial.

````python
cadastro_exemplo = pd.concat(
    [cadastro_exemplo, novo_registro],
    ignore_index=True,
)

cadastro_exemplo
````

O resultado possui três linhas: duas observações copiadas do conjunto e um registro identificado como fictício. O `DataFrame` original `df` continua com 344 observações.
